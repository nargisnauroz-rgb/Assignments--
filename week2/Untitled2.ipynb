```python
clean_file["Department"] = clean_file["Department"].str.strip()
```

```python
clean_file["Department"] = clean_file["Department"].str.upper()
```

```python
clean_file
```

```python
clean_file["Department"] = clean_file["Department"].replace("cs","CS")
```

```python
clean_file["Age"] = clean_file["Age"].replace(250, np.nan)
```

```python
clean_file
```

```python
clean_file["Course_Fee"] = clean_file["Course_Fee"].str.replace("Rs.","")
clean_file["Course_Fee"] = clean_file["Course_Fee"].str.replace("," , "")
```

```python
clean_file
```

```python
clean_file["Course_Fee"] = clean_file["Course_Fee"].replace("not recorded",np.nan)
```

```python
clean_file
```

```python
clean_file["Course_Fee"] = pd.to_numeric(clean_file["Course_Fee"], errors = "coerce")
```
